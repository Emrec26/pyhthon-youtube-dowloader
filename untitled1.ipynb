```python
# -*- coding: utf-8 -*-
"""
YouTube Video & Audio Downloader
Author: [Emre26]
GitHub: [https://github.com/Emrec26]
"""

import os
import sys

try:
    from yt_dlp import YoutubeDL
except ImportError:
    print("[!] Hata: 'yt-dlp' kütüphanesi bulunamadı.")
    print("[!] Lütfen terminalde 'pip install yt-dlp' komutunu çalıştırarak kütüphaneyi yükleyin.\n")
    sys.exit(1)

def media_indir(url, format_secimi="1"):
    """
    Belirtilen YouTube URL'sinden video veya ses indirir.
    """
    kayit_klasoru = "Indirilen_Medyalar"
    if not os.path.exists(kayit_klasoru):
        os.makedirs(kayit_klasoru)

    # Temel indirme ayarları
    ydl_opts = {
        'outtmpl': f'{kayit_klasoru}/%(title)s.%(ext)s',
        'noplaylist': True,  # Yanlışlıkla çalma listesinin tamamını indirmeyi önler
    }

    if format_secimi == "1":
        # En iyi video ve ses kalitesini otomatik birleştirir
        ydl_opts['format'] = 'bestvideo+bestaudio/best'
    elif format_secimi == "2":
        # Sadece ses indirir ve MP3 formatına dönüştürür
        ydl_opts['format'] = 'bestaudio/best'
        ydl_opts['postprocessors'] = [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '192',
        }]
    else:
        print("[X] Geçersiz seçim yapıldı. Varsayılan olarak Video (MP4) indiriliyor.")
        ydl_opts['format'] = 'bestvideo+bestaudio/best'

    try:
        print("\n[+] İndirme başlatılıyor, lütfen bekleyin...")
        with YoutubeDL(ydl_opts) as ydl:
            ydl.download([url])
        print(f"\n[✓] Başarılı! Dosya '{kayit_klasoru}' klasörüne kaydedildi.")
    except Exception as e:
        print(f"\n[X] İndirme sırasında bir hata oluştu: {e}")

def main():
    print("=========================================")
    print("    YouTube Video & MP3 Downloader       ")
    print("=========================================")

    video_url = input("[?] İndirmek istediğiniz YouTube URL'sini yapıştırın: ").strip()

    if not video_url:
        print("[X] Hata: Geçerli bir URL girmediniz.")
        return

    print("\n[1] Video (En Yüksek Kalite)")
    print("[2] Sadece Ses (MP3 Formatı - FFmpeg gerektirir)")
    secim = input("[?] Tercihinizi seçin (1 veya 2): ").strip()

    media_indir(video_url, secim)

if __name__ == "__main__":
    main()
```
